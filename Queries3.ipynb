```r
shopping |> 
  select(`Order ID`, `Order Date`, `Customer Name`) |> 
  left_join(orders |> select(`Order ID`, `Sales`, `Quantity`), by='Order ID') -> shopping2
```

```r
shopping2 |> 
  select(`Customer Name`, `Order Date`, `Order ID`, `Sales`, `Quantity`) |> 
  arrange(`Customer Name`, `Order Date`) |> 
  group_by(`Customer Name`, `Order ID`) |>
  summarise(
    Order_Date = min(`Order Date`), 
    Sales = sum(Sales), 
    Quantity = sum(Quantity)
  ) |> 
  ungroup() |> 
  arrange(`Customer Name`, Order_Date) |> 
  group_by(`Customer Name`) |>
  mutate(sn = row_number()) |> 
  mutate(prev_sale = lag(Sales)) |>
  mutate(status = ifelse(Sales > prev_sale, 'increased', 'reduced')) |>
  #mutate(num_of_sales = max(sn)) |> 
  #mutate(first_sale = first(Sales)) |> 
  #filter(sn == num_of_sales) |>
  ungroup() |>
  head(20)
```

```r
conn <- dbConnect(duckdb())
duckdb_register(conn, "shopping2", shopping2)
```

```r
dbGetQuery(conn, '
SELECT 
  "Order Date", 
  "Customer Name", 
  Sales, 
  --ROW_NUMBER() OVER(PARTITION BY "Customer Name" ORDER BY "Order Date") AS sn 
  --LEAD(Sales) OVER(PARTITION BY "Customer Name" ORDER BY "Order Date") AS sa
  FIRST_VALUE(Sales) OVER(PARTITION BY "Customer Name" ORDER BY "Order Date") AS se
  FROM shopping2
  LIMIT 20

')
```
